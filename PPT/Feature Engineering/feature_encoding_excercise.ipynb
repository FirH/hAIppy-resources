# Latihan: Feature Encoding dengan scikit-learn

Pada latihan ini kamu akan mengubah data kategorikal menjadi angka dengan dua teknik:

| Teknik | Kapan dipakai | Kelas scikit-learn |
|---|---|---|
| **Ordinal Encoding** | kategori punya **urutan** (SMP < SMA < S1 < S2) | `OrdinalEncoder` |
| **One-Hot Encoding** | kategori **tidak punya urutan** (Jakarta, Bandung, Medan) | `OneHotEncoder` |

**Cara mengerjakan:** ganti setiap `____` dengan jawaban yang benar, lalu jalankan sel jawaban dan sel pemeriksa di bagian paling bawah.


## 0. Persiapan — jalankan sel ini terlebih dahulu

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

# Dataset A: survei karyawan (kategori BERURUTAN)
df_ordinal = pd.DataFrame({
    "nama":       ["Andi", "Budi", "Citra", "Dewi", "Eko", "Fajar", "Gita", "Hana"],
    "pendidikan": ["SMA", "S1", "SMP", "S2", "S1", "SMA", "S2", "SMP"],
    "kepuasan":   ["Rendah", "Tinggi", "Sedang", "Tinggi", "Sedang", "Rendah", "Sedang", "Tinggi"],
})

# Dataset B: profil pelanggan (kategori TANPA urutan)
df_nominal = pd.DataFrame({
    "id_pelanggan":  [101, 102, 103, 104, 105, 106, 107, 108],
    "kota":          ["Jakarta", "Bandung", "Surabaya", "Jakarta", "Medan", "Bandung", "Surabaya", "Jakarta"],
    "warna_favorit": ["Merah", "Biru", "Hijau", "Biru", "Merah", "Merah", "Hijau", "Biru"],
})

print("Dataset A - ordinal:")
display(df_ordinal)
print("Dataset B - nominal:")
display(df_nominal)
````

## 2. One-Hot Encoding

Ubah kolom `kota` dan `warna_favorit` pada `df_nominal` menjadi kolom biner 0/1 — satu kolom baru untuk setiap kategori.

## 3. Periksa Jawaban

Jalankan sel di bawah ini setelah kedua soal selesai. Sel ini menghitung ulang jawaban yang benar secara mandiri, lalu membandingkannya dengan hasil milikmu.

````python
def periksa_jawaban():
    lolos = 0
    total = 2

    # ---------- Soal 1: Ordinal Encoding ----------
    print("Soal 1 - Ordinal Encoding")
    try:
        peta_pendidikan = {"SMP": 0, "SMA": 1, "S1": 2, "S2": 3}
        peta_kepuasan   = {"Rendah": 0, "Sedang": 1, "Tinggi": 2}
        benar_pend = df_ordinal["pendidikan"].map(peta_pendidikan).to_numpy(dtype=float)
        benar_kep  = df_ordinal["kepuasan"].map(peta_kepuasan).to_numpy(dtype=float)

        assert "df_ordinal_encoded" in globals(), "variabel `df_ordinal_encoded` belum dibuat."
        for kol in ["pendidikan_enc", "kepuasan_enc"]:
            assert kol in df_ordinal_encoded.columns, f"kolom `{kol}` tidak ditemukan."

        jawab_pend = df_ordinal_encoded["pendidikan_enc"].to_numpy(dtype=float)
        jawab_kep  = df_ordinal_encoded["kepuasan_enc"].to_numpy(dtype=float)

        if not np.array_equal(jawab_pend, benar_pend):
            salah = df_ordinal.loc[jawab_pend != benar_pend, "pendidikan"].unique()
            raise AssertionError(
                f"kolom `pendidikan_enc` belum benar pada kategori {list(salah)}. "
                f"Diharapkan {benar_pend.tolist()}, hasilmu {jawab_pend.tolist()}. "
                "Cek lagi urutan pada `urutan_pendidikan` (paling rendah -> paling tinggi)."
            )
        if not np.array_equal(jawab_kep, benar_kep):
            salah = df_ordinal.loc[jawab_kep != benar_kep, "kepuasan"].unique()
            raise AssertionError(
                f"kolom `kepuasan_enc` belum benar pada kategori {list(salah)}. "
                f"Diharapkan {benar_kep.tolist()}, hasilmu {jawab_kep.tolist()}. "
                "Cek lagi urutan pada `urutan_kepuasan`."
            )

        print("  BENAR - urutan tingkatan sudah tepat.")
        lolos += 1
    except AssertionError as e:
        print(f"  BELUM BENAR - {e}")
    except Exception as e:
        print(f"  ERROR - {type(e).__name__}: {e}")

    # ---------- Soal 2: One-Hot Encoding ----------
    print("\nSoal 2 - One-Hot Encoding")
    try:
        benar = pd.get_dummies(df_nominal[["kota", "warna_favorit"]], prefix_sep="_").astype(int)

        assert "df_onehot_encoded" in globals(), "variabel `df_onehot_encoded` belum dibuat."
        hilang = [k for k in benar.columns if k not in df_onehot_encoded.columns]
        assert not hilang, (
            f"kolom hasil encoding belum lengkap/salah nama: {hilang}. "
            "Pastikan nama kolom diambil dari `get_feature_names_out()`."
        )

        jawab = df_onehot_encoded[list(benar.columns)].astype(int)
        assert jawab.to_numpy().sum() == 2 * len(df_nominal), (
            "setiap baris seharusnya berisi tepat satu angka 1 per kolom asli."
        )
        beda = (jawab.to_numpy() != benar.to_numpy())
        if beda.any():
            baris = np.unique(np.where(beda)[0]).tolist()
            raise AssertionError(f"nilai 0/1 belum sesuai pada baris ke-{baris}.")

        print(f"  BENAR - {len(benar.columns)} kolom biner terbentuk: {list(benar.columns)}")
        lolos += 1
    except AssertionError as e:
        print(f"  BELUM BENAR - {e}")
    except Exception as e:
        print(f"  ERROR - {type(e).__name__}: {e}")

    # ---------- Ringkasan ----------
    print("\n" + "-" * 60)
    print(f"Skor: {lolos}/{total}")
    if lolos == total:
        print("Selamat! Kedua soal encoding sudah benar.")
    else:
        print("Perbaiki bagian `____` yang masih salah, jalankan ulang selnya, lalu periksa lagi.")


periksa_jawaban()
````
