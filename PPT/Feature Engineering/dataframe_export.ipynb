# Export DataFrame ke Gambar

Notebook ini memakai `dataframe_image` (dfi) untuk menyimpan tampilan DataFrame sebagai file gambar, agar bisa langsung dipakai di slide.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
import dataframe_image as dfi

df = pd.DataFrame({
    'size_category': ['small', 'medium', 'large', 'small', 'medium']
})

encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
df['size_encoded'] = encoder.fit_transform(df[['size_category']])
display(df)
```

```python
fruit_data = pd.DataFrame({
    'fruit_category': ['apple', 'orange', 'mango', 'watermelon', 'apple']
})
```

```python
from sklearn.preprocessing import OneHotEncoder
one_hot_encoder = OneHotEncoder(sparse_output=False)
encoded_fruits = one_hot_encoder.fit_transform(fruit_data[['fruit_category']])
encoded_fruit_df = pd.DataFrame(encoded_fruits, columns=one_hot_encoder.get_feature_names_out(['fruit_category']))
fruit_data_encoded = pd.concat([fruit_data, encoded_fruit_df], axis=1)
```

## Export ke PNG

```python
dfi.export(df.head(10), "dataframe.png", table_conversion="matplotlib", dpi=200)
```

```python
dfi.export(fruit_data.head(10), "fruit_data.png", table_conversion="matplotlib", dpi=200)
```

```python
dfi.export(encoded_fruit_df, "encoded_fruit_data.png", table_conversion="matplotlib", dpi=200)
```

```python
dfi.export(fruit_data_encoded.head(10), "combined_fruit_data.png", table_conversion="matplotlib", dpi=200)
```
