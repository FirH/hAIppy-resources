## Fungsi Print
Untuk mendukung anda dalam menampilkan isi sebuah `dataframe` pandas, anda diberikan fungsi yang dapat menerima dataframe untuk dikonversi menjadi list of dictionaries `df_to_dicts` dan melakukan print terhadap list of dictionaries tersebut `print_dicts`

```python
# Dataframe to list of dictionaries converter 
def df_to_dicts(df, columns=None):
    if columns is not None:
        df = df[columns]

    column_names = list(df.columns)
    records = []
    for row in df.itertuples(index=False, name=None):
        records.append(dict(zip(column_names, row)))
    return records

# Print list of dictionaries in a tabular format
def print_dicts(records, columns=None, max_rows=10, max_width=20, missing=""):
    if not records:
        print("(no rows)")
        return

    if columns is None:
        columns = []
        for record in records:
            for key in record:
                if key not in columns:
                    columns.append(key)

    rows_to_show = records if max_rows is None else records[:max_rows]

    def cell(value):
        text = missing if value is None else str(value)
        if len(text) > max_width:
            text = text[: max_width - 1] + "…"
        return text

    header = [cell(name) for name in columns]
    body = [[cell(record.get(name, missing)) for name in columns] for record in rows_to_show]

    widths = [len(text) for text in header]
    for row in body:
        for i, text in enumerate(row):
            widths[i] = max(widths[i], len(text))

    def line(cells):
        return " | ".join(text.ljust(widths[i]) for i, text in enumerate(cells))

    print(line(header))
    print("-+-".join("-" * width for width in widths))
    for row in body:
        print(line(row))

    hidden = len(records) - len(rows_to_show)
    if hidden > 0:
        print(f"... {hidden} more row(s), {len(records)} total")
```

## Demonstrasi Fungsi

### Fungsi Konversi Dataframe

```python
df = pd.read_csv("preprocessed_house.csv")
records = df_to_dicts(df)
display(records)
```

# Fungsi Print Dictionary

```python
print_dicts(records, columns=["id", "price", "bedrooms", "bathrooms", "sqft_living", "yr_built"], max_rows=8)
```
