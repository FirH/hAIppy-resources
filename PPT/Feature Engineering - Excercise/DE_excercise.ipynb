## Fungsi Print
Untuk mendukung anda dalam menampilkan isi sebuah `dataframe` pandas, anda diberikan fungsi yang dapat menerima dataframe untuk dikonversi menjadi list of dictionaries `df_to_dicts` dan melakukan print terhadap list of dictionaries tersebut `print_dicts`

````python
# Dataframe to list of dictionaries converter 
def df_to_dicts(df, columns=None):
    if columns is not None:
        df = df[columns]

    column_names = list(df.columns)
    records = []
    for row in df.itertuples(index=False, name=None):
        records.append(dict(zip(column_names, row)))
    return records

# Print list of dictionaries in a tabular format
def print_dicts(records, columns=None, max_rows=10, max_width=20, missing=""):
    if not records:
        print("(no rows)")
        return

    if columns is None:
        columns = []
        for record in records:
            for key in record:
                if key not in columns:
                    columns.append(key)

    rows_to_show = records if max_rows is None else records[:max_rows]

    def cell(value):
        text = missing if value is None else str(value)
        if len(text) > max_width:
            text = text[: max_width - 1] + "…"
        return text

    header = [cell(name) for name in columns]
    body = [[cell(record.get(name, missing)) for name in columns] for record in rows_to_show]

    widths = [len(text) for text in header]
    for row in body:
        for i, text in enumerate(row):
            widths[i] = max(widths[i], len(text))

    def line(cells):
        return " | ".join(text.ljust(widths[i]) for i, text in enumerate(cells))

    print(line(header))
    print("-+-".join("-" * width for width in widths))
    for row in body:
        print(line(row))

    hidden = len(records) - len(rows_to_show)
    if hidden > 0:
        print(f"... {hidden} more row(s), {len(records)} total")
````

## Fungsi Pengujian
Setiap soal dilengkapi dengan **output yang diharapkan** dan **3 test case**. Jalankan sel test case
setelah mengisi fungsi `answer` untuk meriksa jawaban. Fungsi `is_null` disediakan untuk
membantu memeriksa nilai kosong (`None` atau `NaN`).

Catatan: fungsi `answer` harus mengembalikan **list of dictionaries**, dan urutan baris tidak
diperhitungkan saat pengujian.


````python
import math

# Memeriksa nilai kosong
def is_null(value):
    return value is None or (isinstance(value, float) and math.isnan(value))

# Menjalankan sekumpulan test case terhadap fungsi answer
def run_tests(answer_fn, cases):
    def clean(value):
        if is_null(value):
            return None
        if isinstance(value, float) and float(value).is_integer():
            return int(value)
        return value

    def norm(records):
        rows = [tuple(sorted((k, clean(v)) for k, v in r.items())) for r in records]
        return sorted(rows, key=repr)

    passed = 0
    for number, (title, args, expected) in enumerate(cases, 1):
        try:
            result = answer_fn(*args)
        except Exception as error:
            print(f"[{number}] {title}: ERROR -> {type(error).__name__}: {error}")
            continue
        if norm(result) == norm(expected):
            print(f"[{number}] {title}: PASS")
            passed += 1
        else:
            print(f"[{number}] {title}: FAIL")
            print(f"    diharapkan : {expected}")
            print(f"    hasil anda : {result}")
    print(f"\n{passed}/{len(cases)} test case lulus")
````
